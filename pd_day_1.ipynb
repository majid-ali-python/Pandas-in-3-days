```python
import pandas as pd
x = [3,4,5,6,7]
var = pd.Series(x)
print(var)
```

# How to delete and insert data

```python
#Insert
var = pd.DataFrame({
    "A":[1,2,3,4,5],
    "B":[9,8,7,6,5],
    "C":["M","A","J","I","D"]
})
var.insert(1,"python",var["A"])
var
```

```python
var.insert(2,"Majid",var["C"])
var
```

```python
# Delete
var1 = pd.DataFrame({
    "Majid":["M","A","J","I","D"],
    "Ahmad":["A","H","M","A","D"],
    "Awais":["A","W","A","I","S"]
})
var1.insert(1,"python",[11,12,13,14,15])
var1.pop("Awais")
print(var1)

```
