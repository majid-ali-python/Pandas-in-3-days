```python
#Numeric values ka mean, median ,mode nikalna ho ta describe use karty hai
X.describe()
```

```python
#Data condition ko check karna jasy k jin ki value 70,000 ho un ko print karo
high_salary =X[X["Salary"]>70000]
high_salary
```

```python
# Missing value ko check karna or un ko bharna 
#pely check karna missing vales ko
X.isna().sum()
```

```python
#missing values ko bharna 
#missiong values ko avarage salary say fill karna
avg_salary = X["Salary"].mean()
X["Salary"] =X["Salary"].fillna(avg_salary)
print(avg_salary)
```

```python
# Catagories ko count karna
X.count()
```

```python
# Data grouping k lia 
# department wise average salary nikalne k liya
X.groupby("Department")["Salary"].mean()
```

```python
# Data Sorting
# data ko salary k hisab say highest_t0_Lowest arrange karna
X.sort_values(by="Salary",ascending=False)
```
